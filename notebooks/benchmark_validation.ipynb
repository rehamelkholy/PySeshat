```python
import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from validation.benchmark_runner import (
    run_benchmark
)

from validation.comparison import (
    comparison_table
)

from pyseshat.visualization.orbit_plot import (
    plot_visual_orbit
)
```

## Benchmark Dataset Registry

```python
ROOT = Path("../validation")

BENCHMARK_DIR = (
    ROOT / "benchmark_systems"
)

EXPECTED_DIR = (
    ROOT / "expected_results"
)
```

## Available Benchmark Systems

```python
benchmark_systems = {

    "FIN379": {
        "data_file":
            BENCHMARK_DIR
            / "FIN379"
            / "FIN379.inp",

        "expected_file":
            EXPECTED_DIR
            / "FIN379.json"
    },

    "GL765": {
        "data_file":
            BENCHMARK_DIR
            / "GL765"
            / "GL765.inp",

        "expected_file":
            EXPECTED_DIR
            / "GL765.json"
    },

    "ZetaAqr": {
        "data_file":
            BENCHMARK_DIR
            / "ZetaAqr"
            / "ZetaAqr.in2",

        "expected_file":
            EXPECTED_DIR
            / "ZetaAqr.json"
    }
}

benchmark_systems
```

## Run Benchmark Validation

```python
results = {}

for system_name, config in benchmark_systems.items():

    print("=" * 60)

    print(system_name)

    print("=" * 60)

    result = run_benchmark(
        system_name,
        config
    )

    results[system_name] = result

    print()

    for k, v in result["statistics"].items():

        print(f"{k:20s} = {v:.6f}")

    print()
```

## Parameter Comparison Tables

```python
for system_name, result in results.items():

    print()

    print("=" * 60)

    print(system_name)

    print("=" * 60)

    display(
        comparison_table(result)
    )
```

## Visual Orbit Validation

```python
for system_name, result in results.items():

    print()

    print("=" * 60)

    print(system_name)

    print("=" * 60)

    parsed = result["parsed"]

    fit_result = result["fit_result"]

    visual_data = parsed["visual_data"]

    plot_visual_orbit(
        visual_data,
        fit_result.elements
    )

    plt.show()
```
