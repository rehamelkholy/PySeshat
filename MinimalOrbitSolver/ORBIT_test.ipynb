```python
from utils import *

# --- LOAD ---
elements, data = parse_orbit_file("FIN379.inp")

# --- TRANSFORM ---
t, X, Y, sx, sy = polar_to_cartesian(data)

# --- INITIAL ---
p0 = build_initial_guess(elements)

# --- FIT ---
params_fit = fit_orbit_xy(t, X, Y, sx, sy, p0)

# --- VALIDATION ---
validate_orbit(t, X, Y, sx, sy, params_fit, build_initial_guess(elements))
```
